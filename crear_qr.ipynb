```python
import qrcode

# url = "https://maqui2404.github.io/PortafolioMarco.github.io/"
url = "https://mneuav1iiiccdinfonor.streamlit.app/"
# Crear objeto QR
qr = qrcode.QRCode(
    version=1,  # control del tamaño (1=21x21, hasta 40)
    error_correction=qrcode.constants.ERROR_CORRECT_H,  # nivel de corrección de errores
    box_size=10,  # tamaño de cada "cajita"
    border=4,  # grosor del borde
)

qr.add_data(url)
qr.make(fit=True)

img = qr.make_image(fill_color="black", back_color="white")

img.save("qr_code_streamlit.png")

print("QR generado y guardado como qr_code.png")
```
